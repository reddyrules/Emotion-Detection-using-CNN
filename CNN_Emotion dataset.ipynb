### Splitting

```python
X = ary_dataset_scale
```

```python
y = ary_label
```

```python
X = X.astype("float32")
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.20,random_state=2)
```

```python
X.shape
```

```python
X_train.shape
```

```python
X_test.shape
```

### Model building

```python
from tensorflow.keras.models import Model
```

```python
from tensorflow.keras.layers import Input, Flatten, Dense, Conv2D, MaxPooling2D
```

```python
ip = Input(shape = (64,64,3))
```

### First Convolutional Layer

```python
con1 = Conv2D(filters= 64, kernel_size= (3,3), strides=(1,1), padding= "valid", activation="relu")(ip)
```

```python
max1 = MaxPooling2D(pool_size=(2,2), padding="valid")(con1)
```

### Second Convolutional Layer

```python
con2 = Conv2D(filters = 32, kernel_size = (3,3),strides = (1,1),padding = "valid",activation = "relu")(max1)
```

```python
max2 = MaxPooling2D(pool_size=(2,2),padding="valid")(con2)
```

### Fully Connected Layers

```python
flat = Flatten()(max2)
```

```python
den1 = Dense(128,activation="relu")(flat)
```

```python
den2 = Dense(256,activation="relu")(den1)
```

```python
den3 = Dense(512,activation="relu")(den2)
```

```python
out = Dense(10,activation="softmax")(den3)
```

```python
Model(inputs = ip,outputs= out)
```

```python
cnn_model =Model(inputs = ip,outputs= out) 
```

```python
cnn_model.summary()
```

```python
cnn_model.compile(
    optimizer="adam",
    loss = "sparse_categorical_crossentropy",
    metrics = ["accuracy"]
)
```

```python
cnn_model.fit(X_train,y_train,epochs= 5)
```

```python
cnn_model.evaluate(X_test,y_test)
```
